# 07 - Anomaly Detection

This notebook identifies farms that deviate from typical cluster structure. It combines three unsupervised anomaly signals: distance from cluster centroid, Isolation Forest score, and low weighted degree in the similarity graph.

```python
from pathlib import Path
import json
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.anomaly_detection import run_anomaly_detection

DATA_PATH = PROJECT_ROOT / "data" / "paddydataset.csv"
PROCESSED_DIR = PROJECT_ROOT / "outputs" / "processed"
TABLE_DIR = PROJECT_ROOT / "outputs" / "tables"
FIGURE_DIR = PROJECT_ROOT / "outputs" / "figures"
```

## Run Anomaly Detection

The combined anomaly score uses the following weighting:

- 45% distance from assigned cluster centroid
- 45% Isolation Forest score
- 10% low weighted graph degree

```python
result = run_anomaly_detection(
    data_path=DATA_PATH,
    coordinates_path=PROCESSED_DIR / "pca_coordinates_all_components.csv",
    labels_path=PROCESSED_DIR / "baseline_cluster_labels.csv",
    graph_degree_path=TABLE_DIR / "similarity_graph_degrees.csv",
    output_dir=PROCESSED_DIR,
    table_dir=TABLE_DIR,
    figure_dir=FIGURE_DIR,
    n_components=6,
    cluster_column="selected_baseline_cluster",
    contamination=0.05,
    top_n=50,
    random_state=42,
)

result.metadata
```
