# 05 - Baseline Clustering Comparison

This notebook compares spectral clustering against standard k-means and hierarchical Ward clustering on the same 6-component PCA feature space.

```python
from pathlib import Path
import json
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from src.baseline_clustering import run_baseline_workflow

PROCESSED_DIR = PROJECT_ROOT / "outputs" / "processed"
TABLE_DIR = PROJECT_ROOT / "outputs" / "tables"
FIGURE_DIR = PROJECT_ROOT / "outputs" / "figures"
```

## Run Baselines

The comparison uses the same candidate range as spectral clustering: `k=2` through `k=10`.

```python
result = run_baseline_workflow(
    pca_coordinates_path=PROCESSED_DIR / "pca_coordinates_all_components.csv",
    pca_2d_path=PROCESSED_DIR / "pca_coordinates_2d.csv",
    spectral_metrics_path=TABLE_DIR / "spectral_clustering_metrics.csv",
    output_dir=PROCESSED_DIR,
    table_dir=TABLE_DIR,
    figure_dir=FIGURE_DIR,
    n_components=6,
    min_k=2,
    max_k=10,
    random_state=42,
)

result.metadata
```

## Baseline Metrics

Higher silhouette and Calinski-Harabasz scores are better. Lower Davies-Bouldin scores are better.
